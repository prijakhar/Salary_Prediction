#Preprocessing and model training

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/salaries.csv")
```

```python
df_clean = df.copy()
```

```python
X = df_clean.drop(
    columns=[
        "salary",
        "salary_currency",
        "salary_in_usd"
    ]
)

y = df_clean["salary_in_usd"]
```

```python
X.head()
```

```python
y.head()
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
X_encoded = pd.get_dummies(X, drop_first=True)
```

```python
X_encoded.head()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded,
    y,
    test_size=0.2,
    random_state=42
)
```

```python
print(X_train.shape)
print(X_test.shape)

print(y_train.shape)
print(y_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)
```

```python
print(X_train_scaled.shape)
print(X_test_scaled.shape)
```

```python
X.shape
```

```python
X_encoded.shape
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
X_train_scaled.shape
```

# 10.Machine Learning Phase

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
```

# Train Linear Regression model

```python
linear_model = LinearRegression()
```

```python
linear_model.fit(X_train_scaled, y_train)
```

```python
linear_predictions = linear_model.predict(X_test_scaled)
```

```python
linear_predictions
```

```python
comparison = pd.DataFrame({
    "Actual Salary": y_test,
    "Predicted Salary": linear_predictions
})

comparison
```

# Train decision tree

```python
decision_tree = DecisionTreeRegressor(random_state=42)
```

```python
decision_tree.fit(X_train, y_train)
```

```python
tree_predictions = decision_tree.predict(X_test)
```

```python
tree_predictions
```

# Train Random Forest

```python
random_forest = RandomForestRegressor(
    random_state=42
)
```

```python
random_forest.fit(X_train, y_train)
```

```python
rf_predictions = random_forest.predict(X_test)
```

```python
rf_predictions
```

# Model Evaluation

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import numpy as np
```

#Evaluate Linear regression

```python
linear_mae = mean_absolute_error(y_test, linear_predictions)

linear_mse = mean_squared_error(y_test, linear_predictions)

linear_rmse = linear_mse ** 0.5

linear_r2 = r2_score(y_test, linear_predictions)
```

```python
print("Linear Regression")

print("MAE :", linear_mae)

print("RMSE:", linear_rmse)

print("R2 :", linear_r2)
```

#evaluate decision tree

```python
tree_mae = mean_absolute_error(y_test, tree_predictions)

tree_mse = mean_squared_error(y_test, tree_predictions)

tree_rmse = tree_mse ** 0.5

tree_r2 = r2_score(y_test, tree_predictions)
```

```python
print("Decision Tree")

print("MAE :", tree_mae)

print("RMSE:", tree_rmse)

print("R2 :", tree_r2)
```

#evaluate random forest

```python
rf_mae = mean_absolute_error(y_test, rf_predictions)

rf_mse = mean_squared_error(y_test, rf_predictions)

rf_rmse = rf_mse ** 0.5

rf_r2 = r2_score(y_test, rf_predictions)
```

```python
print("Random Forest")

print("MAE :", rf_mae)

print("RMSE:", rf_rmse)

print("R2 :", rf_r2)
```

#compare all models

```python
results = pd.DataFrame({

    "Model": [
        "Linear Regression",
        "Decision Tree",
        "Random Forest"
    ],

    "MAE": [
        linear_mae,
        tree_mae,
        rf_mae
    ],

    "RMSE": [
        linear_rmse,
        tree_rmse,
        rf_rmse
    ],

    "R2 Score": [
        linear_r2,
        tree_r2,
        rf_r2
    ]

})

results
```

```python
results.sort_values(
    by="R2 Score",
    ascending=False
)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.bar(results["Model"], results["R2 Score"])

plt.title("Model Comparison using R² Score")

plt.xlabel("Models")

plt.ylabel("R² Score")

plt.show()
```

# Step 17 - Retraining Model for Streamlit Deployment

```python
# 1 Select only required columns

df_model = df[[
    "experience_level",
    "employment_type",
    "job_title",
    "company_size",
    "remote_ratio",
    "salary_in_usd"
]]

df_model.head()
```

```python
# 17.2 sepearte features and target
X = df_model.drop("salary_in_usd", axis=1)

y = df_model["salary_in_usd"]
```

```python
print(X.shape)
print(y.shape)
```

```python
# one hot encoding
X_encoded = pd.get_dummies(X)

X_encoded.head()
```

```python
# Step 17.4 Train-Test Splitfrom sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_encoded,
    y,
    test_size=0.2,
    random_state=42
)
```

```python
#17.5 Standard Scaling
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)
```

```python
#17.6 Train the Linear Regression Model
from sklearn.linear_model import LinearRegression

model = LinearRegression()

model.fit(X_train_scaled, y_train)
```

```python
#17.7 Evaluate the Model
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

predictions = model.predict(X_test_scaled)

print("MAE :", mean_absolute_error(y_test, predictions))

print("RMSE :", mean_squared_error(y_test, predictions)**0.5)

print("R2 Score :", r2_score(y_test, predictions))
```

```python
#17.8 Save Everything Again
import joblib

joblib.dump(model, "../models/salary_prediction_model.pkl")

joblib.dump(scaler, "../models/scaler.pkl")

joblib.dump(
    X_encoded.columns.tolist(),
    "../models/model_columns.pkl"
)

print("Model Saved Successfully!")
```
